## 1. Two ways to define $f(x) = x^2$

Python has two ways to define a function:

- `def` creates a **named** function that can contain several lines.
- `lambda` creates a short function made of **one expression**.

```python
def squared(x):
    return x * 2
```

```python
squared_lambda = lambda x: x ** 2
```

**Question 1.** Do the two functions agree at $x = 2$?

```python
print(squared(2), squared_lambda(2))
```

**Question 2.** Both print `4`. Does that mean both functions are correct? Plot them on $[-3, 3]$ and look.

```python
grid = np.linspace(-3, 3, 61)

plt.plot(grid, squared(grid), label="squared (def)")
plt.plot(grid, squared_lambda(grid), label="squared_lambda (lambda)", linestyle="--")
plt.legend()
plt.title("Do these two functions really compute the same thing?")
plt.show()
```
