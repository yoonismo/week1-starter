## 1. Two ways to define $f(x) = x^2$

Python has two ways to define a function:

- `def` creates a **named** function that can contain several lines.
- `lambda` creates a short function made of **one expression**.

```python
def squared(x):
    return x * 2
```

```python
squared_lambda = lambda x: x ** 2
```
